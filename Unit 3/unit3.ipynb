```python
# vector Quantization

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from skimage import io
# pip install scikit-image 

#reading an image
img = io.imread('i1.jpeg')
plt.imshow(img)
plt.title('source image')
plt.show()

#resharping the image data
pixels = img.reshape(-1,3)
# converts image to 2D array , one pixel per row , 3 columns for RGB

# applying the k-Means for color quantization
k = 16
kmeans = KMeans(n_clusters = k)
kmeans.fit(pixels)
new_color = kmeans.cluster_centers_[kmeans.labels_]
# clustyer_centers_ , gives representative colors
# labels_tells which cluster each pixel belongs to

#  Recreating the compressed image
compressed = new_color.reshape(img.shape).astype('uint8')
# 8- bit unsigned integer is a standard datatype for image pixels

# display
plt.imshow(compressed)
plt.title('compressed images')
plt.show()
```

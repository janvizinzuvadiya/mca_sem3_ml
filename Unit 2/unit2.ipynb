```python
# UNIT 2 
#  Pre - Processing

# Mean Removal

import  numpy as np
a = np.array([[2],[4],[6],[8],[10]])

print(a)

mean = np.mean(a)
print(mean)

cen = a-mean
print(cen)
```

```python
# Scaling

import numpy as np 
a = np.array([[10],[20],[30],[40],[50]])
print(a)

min = np.min(a)
print(min)

max = np.max(a)
print(max)

scal = (a-min)/(max-min)
print(scal)
```

```python
# pre processing
# 3. normalization

import numpy as np
from sklearn import preprocessing
# scikit learn

data = ([[3,4],[1,2],[5,12]])
print(data)

# L - lebesgue space
nl1 = preprocessing.normalize(data,norm='l1')
print(nl1)

nl2 = preprocessing.normalize(data,norm='l2')
print(nl2)
```

```python
# pre processing
#binarization

import numpy as np
from sklearn import preprocessing

data = np.array([[5],[7],[12],[15],[25]])
print(data)
print('*'*15)

bin = preprocessing.Binarizer(threshold=7).transform(data)
#  applies binarizer to each element
print(bin)
```

```python
# pre processng 
# label endoding
from sklearn import preprocessing

label  = ['apple','mango','cherry','dragon fruit','grapes']
print(label)

# load the encoder
lblen = preprocessing.LabelEncoder()

# fit and transform
op = lblen.fit_transform(label)
print(op)
```

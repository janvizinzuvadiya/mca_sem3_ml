```python
# Different types od bluring effect

import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('i2.png')
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('m.jpeg')
img = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

# sobel (source_image , ddepth , x,y , kernel_size)
#  64 bit floating point value - CV_64F
# 1,0,5 -> horizontal , vertical , kernelsize

sobelx = cv2.Sobel(img , cv2.CV_64F , 1,0,5)
sobely = cv2.Sobel(img , cv2.CV_64F , 0,1,5)
sobel_combined = cv2.magnitude(sobelx , sobely) 

plt.subplot(1,4,1),plt.imshow(img),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,4,2),plt.imshow(sobelx, cmap = 'gray'),plt.title('Horizontal Sobel'),plt.axis('off')
plt.subplot(1,4,3),plt.imshow(sobely, cmap = 'gray'),plt.title('Vertical Sobel'),plt.axis('off')
plt.subplot(1,4,4),plt.imshow(sobel_combined, cmap = 'gray'),plt.title('Sobel'),plt.axis('off')

# Box Blur :
box = cv2.blur(img,(9,9))

# Gaussian blur
# must be odd value for dimentions of kernel
gs = cv2.GaussianBlur(img,(5,5),0)

# Median Blur
# must be odd value for dimentions of kernel , bigger than 1 , any positive num
md = cv2.medianBlur(img,5)

plt.subplot(2,2,1),plt.imshow(img),plt.axis('off'),plt.title('Source Image')
plt.subplot(2,2,2),plt.imshow(box),plt.axis('off'),plt.title('Box Blur Image')
plt.subplot(2,2,1),plt.imshow(gs),plt.axis('off'),plt.title('Gaussian Blur Image')
plt.subplot(2,2,2),plt.imshow(md),plt.axis('off'),plt.title('Median Blur Image')
```

```python
#  Motion Blur


import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('i2.png')
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
size = 30

k1 = np.zeros((size,size))
k1[int(size/2),:] = np.ones(size)
k1 = k1/size
k1_horizontal = cv2.filter2D(img,-1,k1)

k2 = np.zeros((size,size))
k2[:,int(size/2)] = np.ones(size)
k2 = k2/size
k2_vertical = cv2.filter2D(img,-1,k2)

plt.subplot(1,3,1),plt.imshow(img),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,3,2),plt.imshow(k1_horizontal),plt.title('Horizontal Blur'),plt.axis('off')
plt.subplot(1,3,3),plt.imshow(k2_vertical),plt.title('Vertical Blur'),plt.axis('off')
```

```python
# Edge Detection
# 2 Laplacian

import cv2
import matplotlib.pyplot as plt

img = cv2.imread('m.png')
img = cv2.cvtColor(img , cv2.COLOR_BGR2GRAY)

#Gaussian blur
gaus = cv2.GaussianBlur (img, (5,5),0)

# Laplacian ( src_img , ddepth , k_size )
lap = cv2.Laplacian (gaus , cv2.CV_64F , 3 )

# used to adjust brightness and constrast  generally to an 8- bit   
lap = cv2.convertScaleAbs(lap)

plt.subplot(1,2,1),plt.imshow(img),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,2,2),plt.imshow(lap,cmap = 'gray'),plt.title('Laplacian Image'),plt.axis('off')

```

```python
# Edge Detection
# 3 Canny

import cv2
import matplotlib.pyplot as plt

img = cv2.imread('m.png')
img = cv2.cvtColor(img , cv2.COLOR_BGR2GRAY)

#Gaussian blur
gaus = cv2.GaussianBlur(img, (5,5),0)
            
#Canny (src , threshold1 , threshold2)
cann = cv2.Canny(gaus,120,200)

plt.subplot(1,2,1),plt.imshow(img),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,2,2),plt.imshow(cann,cmap = 'gray'),plt.title('Canny Image'),plt.axis('off')
```

```python
# erosion

import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('ero.jpeg')
img = cv2.bitwise_not(img)

k = np.ones((3,3),np.uint8)

e = cv2.erode(img, k , iterations = 15)
d = cv2.dilate(img , k ,iterations = 15)

plt.subplot(1,3,1),plt.imshow(img, cmap = "grey"),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,3,2),plt.imshow(e, cmap = "grey"),plt.title('Erode'),plt.axis('off')
plt.subplot(1,3,3),plt.imshow(d, cmap = "grey"),plt.title('Dilation'),plt.axis('off')
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('m.png')

# convert to ycrcb
ycrcb = cv2.cvtColor(img , cv2.COLOR_BGR2YCrCb)
y, cr , cb = cv2.split(ycrcb)

# applying clahe
clahe = cv2.createCLAHE(clipLimit = 2.0 , tileGridSize = (8,8))
y_clahe = clahe.apply(y)
ycrcb = cv2.merge([y_clahe, cr, cb])
clahe_eq = cv2.cvtColor(ycrcb , cv2.COLOR_YCrCb2BGR)

plt.figure(figsize = (6,5))

plt.subplot(1,2,1),plt.imshow(cv2.cvtColor(img , cv2.COLOR_BGR2RGB)),plt.title('Source Image'),plt.axis('off')           
plt.subplot(1,2,2),plt.imshow(cv2.cvtColor(clahe_eq , cv2.COLOR_BGR2RGB)),plt.title('Contrast Image'),plt.axis('off')           
```

```python
# advanced image filtering
# 2. Advanced Sharping using Unsharp

import cv2
import matplotlib.pyplot as plt

img = cv2.imread('ct.jpeg')
gs = cv2.GaussianBlur(img,(9,9),10)

#unsharp
# (source_imagex1.5) + (blured_imagex-0.5)
unsharp = cv2.addWeighted(img , 1.5 , gs , -0.5 , 0) 

plt.subplot(1,2,1),plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB)),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,2,2),plt.imshow(cv2.cvtColor(unsharp,cv2.COLOR_BGR2RGB)),plt.title('Unsharp Filter Image'),plt.axis('off')
```

```python
# Advanced image filters
# Edge detection using scharr

import cv2
import matplotlib.pyplot as plt

img = cv2.imread('fl.jpeg',0)

#Scharr
scx = cv2.Scharr(img,cv2.CV_64F,1,0)
scy = cv2.Scharr(img,cv2.CV_64F,0,1)

# Convert to uint8

scx = cv2.convertScaleAbs(scx)
scy = cv2.convertScaleAbs(scy)

# Combine
sc = cv2.addWeighted(scx , 0.10 , scy , 0.2 , 0)

plt.subplot(1,2,1),plt.imshow(img,cmap = 'gray'),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,2,2),plt.imshow(sc, cmap = 'gray'),plt.title('Source Image'),plt.axis('off')

```

```python
# object detection
# face from the image

import cv2

# load the pre-trained face detector
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

# Read the Image
img = cv2.imread('faces.jpeg')

#Read the image
# haas cascade works on single channel color
gray_image = (cv2.cvtColor(img,cv2.COLOR_BGR2GRAY))

faces = face_cascade.detectMultiScale(gray_image,scaleFactor = 1.1, minNeighbors = 4)
print(faces)

for (x,y,w,h) in faces:
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),3)

cv2.imshow('Face Detection' , img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# 1. Load the pre-trained face detector
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

# 2. Open webcam (0 = default camera)
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    
    # If camera feed is unavailable or interrupted
    if not ret:
        print('Cannot access camera feed')
        break

    # 3. Convert frame to GRAYSCALE for Haar Cascade
    gray_image = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # 4. Detect faces in the current frame
    faces = face_cascade.detectMultiScale(gray_image, scaleFactor=1.1, minNeighbors=4)

    # 5. Draw bounding boxes on 'frame' (not 'img')
    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 3)

    # 6. Display the live video stream with bounding boxes
    cv2.imshow('Face Detection from Camera', frame)

    # 7. Press 'ESC' key (ASCII 27) or 'q' to exit loop
    if cv2.waitKey(1) & 0xFF == 27:
        break

# --- Cleanup MUST be outside the loop ---
cap.release()
cv2.destroyAllWindows()
```

```python
# Different types od bluring effect

import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('i2.png')
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('m.jpeg')
img = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

# sobel (source_image , ddepth , x,y , kernel_size)
#  64 bit floating point value - CV_64F
# 1,0,5 -> horizontal , vertical , kernelsize

sobelx = cv2.Sobel(img , cv2.CV_64F , 1,0,5)
sobely = cv2.Sobel(img , cv2.CV_64F , 0,1,5)
sobel_combined = cv2.magnitude(sobelx , sobely)

plt.subplot(1,4,1),plt.imshow(img),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,4,2),plt.imshow(sobelx, cmap = 'gray'),plt.title('Horizontal Sobel'),plt.axis('off')
plt.subplot(1,4,3),plt.imshow(sobely, cmap = 'gray'),plt.title('Vertical Sobel'),plt.axis('off')
plt.subplot(1,4,4),plt.imshow(sobel_combined, cmap = 'gray'),plt.title('Sobel'),plt.axis('off')


# Box Blur :
box = cv2.blur(img,(9,9))

# Gaussian blur
# must be odd value for dimentions of kernel
gs = cv2.GaussianBlur(img,(5,5),0)

# Median Blur
# must be odd value for dimentions of kernel , bigger than 1 , any positive num
md = cv2.medianBlur(img,5)


plt.subplot(2,2,1),plt.imshow(img),plt.axis('off'),plt.title('Source Image')
plt.subplot(2,2,2),plt.imshow(box),plt.axis('off'),plt.title('Box Blur Image')
plt.subplot(2,2,1),plt.imshow(gs),plt.axis('off'),plt.title('Gaussian Blur Image')
plt.subplot(2,2,2),plt.imshow(md),plt.axis('off'),plt.title('Median Blur Image')
```

```python
#  Motion Blur


import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('i2.png')
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
size = 30

k1 = np.zeros((size,size))
k1[int(size/2),:] = np.ones(size)
k1 = k1/size
k1_horizontal = cv2.filter2D(img,-1,k1)

k2 = np.zeros((size,size))
k2[:,int(size/2)] = np.ones(size)
k2 = k2/size
k2_vertical = cv2.filter2D(img,-1,k2)

plt.subplot(1,3,1),plt.imshow(img),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,3,2),plt.imshow(k1_horizontal),plt.title('Horizontal Blur'),plt.axis('off')
plt.subplot(1,3,3),plt.imshow(k2_vertical),plt.title('Vertical Blur'),plt.axis('off')
```

```python
# Edge Detection
# 2 Laplacian

import cv2
import matplotlib.pyplot as plt

img = cv2.imread('m.png')
img = cv2.cvtColor(img , cv2.COLOR_BGR2GRAY)

#Gaussian blur
gaus = cv2.GaussianBlur (img, (5,5),0)

# Laplacian ( src_img , ddepth , k_size )
lap = cv2.Laplacian (gaus , cv2.CV_64F , 3 )

# used to adjust brightness and constrast  generally to an 8- bit   
lap = cv2.convertScaleAbs(lap)

plt.subplot(1,2,1),plt.imshow(img),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,2,2),plt.imshow(lap,cmap = 'gray'),plt.title('Laplacian Image'),plt.axis('off')

```

```python
# Edge Detection
# 3 Canny

import cv2
import matplotlib.pyplot as plt

img = cv2.imread('m.png')
img = cv2.cvtColor(img , cv2.COLOR_BGR2GRAY)

#Gaussian blur
gaus = cv2.GaussianBlur(img, (5,5),0)
            
#Canny (src , threshold1 , threshold2)
cann = cv2.Canny(gaus,120,200)

plt.subplot(1,2,1),plt.imshow(img),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,2,2),plt.imshow(cann,cmap = 'gray'),plt.title('Canny Image'),plt.axis('off')
```

```python


# Embossing Image

import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('ei4.jpeg',cv2.IMREAD_GRAYSCALE)

k1 = np.array([[0,-1,-1],[1,0,-1],[1,1,0]])
img1 = cv2.filter2D(img,-1,k1)

k2 = np.array([[-1,-1,0],[-1,0,1],[0,1,1]])
img2 = cv2.filter2D(img,-1,k2)

k3 = np.array([[1,0,0],[0,0,0],[0,0,-1]])
img3 = cv2.filter2D(img,-1,k3)

k4 = np.array([[-2,-1,0],[-1,1,1],[0,1,2]])
img4 = cv2.filter2D(img,-1,k4)


plt.subplot(2,3,1),plt.imshow(img),plt.title('source Image'),plt.axis('off')
plt.subplot(2,3,2),plt.imshow(img1,cmap = 'gray'),plt.title('Embosssed Image 1'),plt.axis('off')
plt.subplot(2,3,3),plt.imshow(img2,cmap = 'gray'),plt.title('Embossed Image 2'),plt.axis('off')
plt.subplot(2,3,4),plt.imshow(img3,cmap = 'gray'),plt.title('Embossed Image 3'),plt.axis('off')
plt.subplot(2,3,5),plt.imshow(img4,cmap = 'gray'),plt.title('Embossed Image 4'),plt.axis('off')
```

```python
# erosion

import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('ero.jpeg')
img = cv2.bitwise_not(img)

k = np.ones((3,3),np.uint8)

e = cv2.erode(img, k , iterations = 15)
d = cv2.dilate(img , k ,iterations = 15)

plt.subplot(1,3,1),plt.imshow(img, cmap = "grey"),plt.title('Source Image'),plt.axis('off')
plt.subplot(1,3,2),plt.imshow(e, cmap = "grey"),plt.title('Erode'),plt.axis('off')
plt.subplot(1,3,3),plt.imshow(d, cmap = "grey"),plt.title('Dilation'),plt.axis('off')
```
